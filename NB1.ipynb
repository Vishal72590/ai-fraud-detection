# Class imbalance

```python
# 'Class' column: 0 = legit transaction, 1 = fraud
class_counts = df['Class'].value_counts()
class_pct    = df['Class'].value_counts(normalize=True) * 100

print(f'Legitimate: {class_counts[0]:,}  ({class_pct[0]:.2f}%)')
print(f'Fraud:      {class_counts[1]:,}  ({class_pct[1]:.2f}%)')
print(f'Imbalance ratio: {class_counts[0]//class_counts[1]}:1')

# Plot
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(['Legitimate', 'Fraud'], class_counts.values,
            color=['#2196F3', '#F44336'])
axes[0].set_title('Transaction Count', fontweight='bold')
for i, v in enumerate(class_counts.values):
    axes[0].text(i, v + 500, f'{v:,}', ha='center', fontweight='bold')

axes[1].pie(class_counts.values,
            labels=['Legitimate', 'Fraud'],
            autopct='%1.2f%%',
            colors=['#2196F3', '#F44336'],
            explode=(0, 0.1))
axes[1].set_title('Class Distribution', fontweight='bold')

plt.tight_layout()
plt.show()                                         
```

# Amount vs Fraud

```python
fraud_df = df[df['Class'] == 1]   # Only fraud rows
legit_df = df[df['Class'] == 0]   # Only legit rows

print(f'Fraud  avg amount: ${fraud_df["Amount"].mean():.2f}')
print(f'Legit  avg amount: ${legit_df["Amount"].mean():.2f}')

# Fix: use fig.add_subplot instead of plt.subplots
fig = plt.figure(figsize=(14, 5))

# Left plot — Amount distribution
ax1 = fig.add_subplot(1, 2, 1)
ax1.hist(np.log1p(fraud_df['Amount']), bins=50,
         alpha=0.7, color='#F44336', label='Fraud')
ax1.hist(np.log1p(legit_df['Amount']), bins=50,
         alpha=0.4, color='#2196F3', label='Legit')
ax1.set_title('Amount Distribution (log scale)', fontweight='bold')
ax1.set_xlabel('log(Amount + 1)')
ax1.set_ylabel('Count')
ax1.legend()

# Right plot — Time distribution
ax2 = fig.add_subplot(1, 2, 2)
ax2.hist(fraud_df['Time'] / 3600, bins=48,
         alpha=0.7, color='#F44336', label='Fraud')
ax2.hist(legit_df['Time'] / 3600, bins=48,
         alpha=0.4, color='#2196F3', label='Legit')
ax2.set_title('Time of Transaction (hours)', fontweight='bold')
ax2.set_xlabel('Hours')
ax2.set_ylabel('Count')
ax2.legend()

plt.tight_layout()
plt.show()
```

# Top correlated features

```python
# Correlation: how strongly each feature relates to fraud (Class)
# abs() = we care about strength, not direction

correlation = df.corr()['Class'].abs().sort_values(ascending = False)

print('Top 10 features correlation with fraud')
print(correlation.drop('Class').head(10))

top10 = correlation.drop('Class').head(10)
top10.plot(kind='bar', color='#E91E63', figsize=(10, 5))
plt.title('Top 10 Features vs Fraud', fontweight='bold')
plt.ylabel('Absolute Correlation')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import mysql.connector

conn = mysql.connector.connect(
    host     = 'localhost',
    user     = 'root',
    password = 'Rathod@9vishal'  
)
cursor = conn.cursor()

# Create database
cursor.execute("CREATE DATABASE IF NOT EXISTS fraud_detection_db")
cursor.execute("USE fraud_detection_db")
print('✅ Database created: fraud_detection_db')

# ── Table 1: raw_transactions ──────────────────────────────────────────
cursor.execute("""
CREATE TABLE IF NOT EXISTS raw_transactions (
    id          INT AUTO_INCREMENT PRIMARY KEY,
    Time        FLOAT,
    V1  FLOAT, V2  FLOAT, V3  FLOAT, V4  FLOAT, V5  FLOAT,
    V6  FLOAT, V7  FLOAT, V8  FLOAT, V9  FLOAT, V10 FLOAT,
    V11 FLOAT, V12 FLOAT, V13 FLOAT, V14 FLOAT, V15 FLOAT,
    V16 FLOAT, V17 FLOAT, V18 FLOAT, V19 FLOAT, V20 FLOAT,
    V21 FLOAT, V22 FLOAT, V23 FLOAT, V24 FLOAT, V25 FLOAT,
    V26 FLOAT, V27 FLOAT, V28 FLOAT,
    Amount      FLOAT,
    Class       INT,
    loaded_at   TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    source_file VARCHAR(100)
)
""")
print('✅ Table 1 ready: raw_transactions')

# ── Table 2: processed_transactions ───────────────────────────────────
cursor.execute("""
CREATE TABLE IF NOT EXISTS processed_transactions (
    id            INT AUTO_INCREMENT PRIMARY KEY,
    raw_id        INT,
    V1  FLOAT, V2  FLOAT, V3  FLOAT, V4  FLOAT, V5  FLOAT,
    V6  FLOAT, V7  FLOAT, V8  FLOAT, V9  FLOAT, V10 FLOAT,
    V11 FLOAT, V12 FLOAT, V13 FLOAT, V14 FLOAT, V15 FLOAT,
    V16 FLOAT, V17 FLOAT, V18 FLOAT, V19 FLOAT, V20 FLOAT,
    V21 FLOAT, V22 FLOAT, V23 FLOAT, V24 FLOAT, V25 FLOAT,
    V26 FLOAT, V27 FLOAT, V28 FLOAT,
    Amount_scaled FLOAT,
    amount_log    FLOAT,
    hour_of_day   INT,
    is_night      INT,
    amount_bin    INT,
    Class         INT,
    processed_at  TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (raw_id) REFERENCES raw_transactions(id)
)
""")
print('✅ Table 2 ready: processed_transactions')

# ── Table 3: model_predictions ─────────────────────────────────────────
cursor.execute("""
CREATE TABLE IF NOT EXISTS model_predictions (
    id                INT AUTO_INCREMENT PRIMARY KEY,
    processed_id      INT,
    lr_score          FLOAT,
    rf_score          FLOAT,
    xgb_score         FLOAT,
    iso_score         FLOAT,
    ae_score          FLOAT,
    final_probability FLOAT,
    decision          VARCHAR(10),
    risk_level        VARCHAR(10),
    actual_class      INT,
    predicted_at      TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (processed_id) REFERENCES processed_transactions(id)
)
""")
print('✅ Table 3 ready: model_predictions')

# ── Table 4: model_metrics ─────────────────────────────────────────────
cursor.execute("""
CREATE TABLE IF NOT EXISTS model_metrics (
    id           INT AUTO_INCREMENT PRIMARY KEY,
    model_name   VARCHAR(50),
    pr_auc       FLOAT,
    roc_auc      FLOAT,
    f1_score     FLOAT,
    threshold    FLOAT,
    dataset      VARCHAR(20),
    evaluated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")
print('✅ Table 4 ready: model_metrics')

# ── Table 5: smote_split_log ───────────────────────────────────────────
cursor.execute("""
CREATE TABLE IF NOT EXISTS smote_split_log (
    id           INT AUTO_INCREMENT PRIMARY KEY,
    train_size   INT,
    val_size     INT,
    test_size    INT,
    train_fraud  INT,
    train_legit  INT,
    smote_fraud  INT,
    smote_legit  INT,
    created_at   TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")
print('✅ Table 5 ready: smote_split_log')

# ── Table 6: fraud_alerts ──────────────────────────────────────────────
cursor.execute("""
CREATE TABLE IF NOT EXISTS fraud_alerts (
    id                INT AUTO_INCREMENT PRIMARY KEY,
    prediction_id     INT,
    fraud_probability FLOAT,
    decision          VARCHAR(10),
    amount            FLOAT,
    hour_of_day       INT,
    alert_status      VARCHAR(20) DEFAULT 'OPEN',
    created_at        TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")
print('✅ Table 6 ready: fraud_alerts')

conn.commit()
cursor.close()
conn.close()
print('\n🎉 All 6 tables created in fraud_detection_db!')
print('   You can verify in MySQL Workbench now.')
```

#  Save clean dataset

```python
import mysql.connector

print('Saving raw data to MySQL...')
print('This will take 2-3 minutes (284,807 rows)...')

conn = mysql.connector.connect(
    host     = 'localhost',
    user     = 'root',
    password = 'Rathod@9vishal',    
    database = 'fraud_detection_db'
)
cursor = conn.cursor()

# Clear old data if rerunning this cell
cursor.execute("DELETE FROM raw_transactions")
conn.commit()

# Build column list
v_cols = [f'V{i}' for i in range(1, 29)]
all_cols = ['Time'] + v_cols + ['Amount', 'Class']

# Insert query — %s = placeholder for each value
placeholders = ','.join(['%s'] * (len(all_cols) + 1))  # +1 for source_file
insert_query = f"""
    INSERT INTO raw_transactions
    (Time, {','.join(v_cols)}, Amount, Class, source_file)
    VALUES ({placeholders})
"""

# Insert in batches of 1000 rows — much faster than one by one
BATCH_SIZE = 1000
batch      = []
total      = 0

for _, row in df[all_cols].iterrows():
    batch.append(tuple(row) + ('creditcard.csv',))

    if len(batch) == BATCH_SIZE:
        cursor.executemany(insert_query, batch)
        conn.commit()
        total += BATCH_SIZE
        batch  = []
        print(f'   Inserted {total:,} / {len(df):,} rows...')

# Insert leftover rows
if batch:
    cursor.executemany(insert_query, batch)
    conn.commit()
    total += len(batch)

cursor.close()
conn.close()

print(f'\n Done! {total:,} rows saved to raw_transactions table')
print('   Open MySQL Workbench → fraud_detection_db → raw_transactions to verify!')
```
